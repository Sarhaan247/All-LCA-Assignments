```python
# Create a dictionary with 5 elements
my_dict = {
    "name": "ABC",
    "rollno": 34,
    "branch": "CSE",
    "age": 18,
    "marks": 85
}

print("Original dictionary:")
print(my_dict)

# Add student information
my_dict["city"] = "Pune"

print("\nAfter adding information:")
print(my_dict)

# Delete student information
del my_dict["age"]

print("\nAfter deleting age:")
print(my_dict)

# Display student information
print("\nStudent Information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

values = mydict.values()

mean = sum(values) / len(values)

print("Mean of all values:", mean)
```

```python
my_dict = {
    "name": ["Yash", "Neel", "Dev"],
    "rollno": [11, 12, 13],
    "marks": [78, 56, 98]
}

# a) Display name as Dev
print("Name:", my_dict["name"][2])

# b) Display roll no. 12
print("Roll No:", my_dict["rollno"][1])

# c) Display greatest marks
print("Greatest marks:", max(my_dict["marks"]))
```

```python
string = input("Enter a string: ")

frequency = {}

for letter in string:
    if letter in frequency:
        frequency[letter] += 1
    else:
        frequency[letter] = 1

print("Frequency of letters:")

for letter, count in frequency.items():
    print(letter, ":", count)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Quotient and Remainder:")
print(result)
```

```python
from datetime import date

# Take first date
d1 = int(input("Enter day of first date: "))
m1 = int(input("Enter month of first date: "))
y1 = int(input("Enter year of first date: "))

# Take second date
d2 = int(input("Enter day of second date: "))
m2 = int(input("Enter month of second date: "))
y2 = int(input("Enter year of second date: "))

# Create date tuples
date1 = (d1, m1, y1)
date2 = (d2, m2, y2)

# Convert tuples to date objects
date1_obj = date(y1, m1, d1)
date2_obj = date(y2, m2, d2)

# Find difference
difference = abs((date2_obj - date1_obj).days)

print("First date:", date1)
print("Second date:", date2)
print("Number of days between two dates:", difference)
```
